```python
import matplotlib.pyplot as plt

# Data extraction
periods = [
    "Mitterrand (81–88)",
    "Mitterrand (88–95)",
    "Chirac (95–02)",
    "Chirac (02–07)",
    "Sarkozy (07–12)",
    "Hollande (12–17)",
    "Macron (17–22)",
    "Macron (22–juin 26)"
]

ecart = [203.9, 383.4, 265.2, 302.0, 647.0, 347.5, 692.4, 639.9]

# Create figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Create bar chart
bars = ax.bar(periods, ecart, color='royalblue', edgecolor='darkblue', alpha=0.85)

# Add value labels on top of each bar
for bar in bars:
    height = bar.get_height()
    ax.annotate(f'+{height} Md€',
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 4),  # offset points
                textcoords="offset points",
                ha='center', va='bottom', fontsize=9, fontweight='bold')

# Chart styling
ax.set_ylabel('Augmentation de la dette (en milliards d\'euros)', fontsize=11)
ax.set_title('Évolution de la dette publique par mandat présidentiel en France', fontsize=13, fontweight='bold', pad=15)
plt.xticks(rotation=45, ha='right', fontsize=10)

# Clean up top and right borders
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Layout adjustment
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Timeline matching the presidential term milestones
years = [
    '1988\n(Mitterrand)', 
    '1995\n(Mitterrand)', 
    '2002\n(Chirac)', 
    '2007\n(Chirac)', 
    '2012\n(Sarkozy)', 
    '2017\n(Hollande)', 
    '2022\n(Macron)', 
    '2026\n(Macron)'
]

# Representative macroeconomic data points
pib_nominal = [900, 1220, 1530, 1900, 2080, 2300, 2630, 3600]  # in Billion €
cac40_values = [1500, 2000, 4400, 5600, 3350, 5280, 6470, 8080]  # index points
purchasing_power = [108, 105, 100, 96, 92, 88, 83, 80]  # Base 100 in 2000 (showing erosion)

# Create a figure with 3 stacked subplots sharing the x-axis
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 11), sharex=True)

# 1. Nominal GDP Plot
ax1.plot(years, pib_nominal, marker='o', color='royalblue', linewidth=2.5, markersize=6)
ax1.set_ylabel('PIB Nominal (Md €)', fontsize=10, fontweight='bold')
ax1.set_title('Macroeconomic Comparison Across Presidential Terms in France', fontsize=13, fontweight='bold', pad=15)
ax1.grid(True, linestyle='--', alpha=0.5)

# 2. CAC 40 Market Index Plot
ax2.plot(years, cac40_values, marker='s', color='forestgreen', linewidth=2.5, markersize=6)
ax2.set_ylabel('CAC 40 Index', fontsize=10, fontweight='bold')
ax2.grid(True, linestyle='--', alpha=0.5)

# 3. Public Sector Purchasing Power Index Plot
ax3.plot(years, purchasing_power, marker='^', color='firebrick', linewidth=2.5, markersize=6)
ax3.set_ylabel('Pouvoir d\'achat\n(Indice Base 100 = 2000)', fontsize=10, fontweight='bold')
ax3.set_xlabel('Fin de mandat / Période', fontsize=10, fontweight='bold')
ax3.grid(True, linestyle='--', alpha=0.5)

# Clean up top and right spines for all subplots
for ax in [ax1, ax2, ax3]:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.xticks(fontsize=9)
plt.tight_layout()
plt.show()
```
