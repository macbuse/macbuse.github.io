```python

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Wedge

# Define the figure and axis
fig, ax = plt.subplots(figsize=(4,4))

# Angles in degrees for matplotlib Wedge
theta1 = -60  # -pi/3
theta2 = 90   # pi/2

# Outer sector (r = 3) drawn dashed since |z| < 3 (strict inequality)
outer_wedge = Wedge((0, 0), 3, theta1, theta2, color='skyblue', alpha=0.6)
ax.add_patch(outer_wedge)

# Inner sector (r = 2) drawn in white to punch out the inner circle
inner_wedge = Wedge((0, 0), 2, theta1, theta2, color='white', zorder=2)
ax.add_patch(inner_wedge)

# Boundaries representation
theta_grid = np.linspace(np.radians(theta1), np.radians(theta2), 200)

# Inner boundary (|z| = 2): Solid line
ax.plot(2 * np.cos(theta_grid), 2 * np.sin(theta_grid), color='navy', lw=2)

# Outer boundary (|z| = 3): Dashed line (strict inequality)
ax.plot(3 * np.cos(theta_grid), 3 * np.sin(theta_grid), color='navy', lw=2, linestyle='--')

# Radial boundaries
ax.plot([0, 3 * np.cos(np.radians(theta1))], [0, 3 * np.sin(np.radians(theta1))], color='navy', lw=1.5)
ax.plot([0, 3 * np.cos(np.radians(theta2))], [0, 3 * np.sin(np.radians(theta2))], color='navy', lw=1.5)

# Center point (origin excluded since z != 0)
ax.plot(0, 0, marker='o', mfc='white', mec='navy', zorder=3)

# Configure axes
ax.set_xlim([-4, 4])
ax.set_ylim([-4, 4])
ax.axhline(0, color='black', linewidth=0.8, alpha=0.5)
ax.axvline(0, color='black', linewidth=0.8, alpha=0.5)

# Centered spines
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

plt.grid(True, linestyle=':', alpha=0.6)
plt.title(r' $-\frac{\pi}{3} \leq \arg(z) \leq \frac{\pi}{2}$ and $2 \leq |z| < 3$')

ax.set_aspect('equal', adjustable='box')
plt.tight_layout()

# Save and display
plt.savefig('1_13f.png', dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Wedge

fig, ax = plt.subplots(figsize=(5,5))

# -------------------------------------------------------------------------
# 1. Whole Annulus: 2 <= |z| < 3 (Light Blue)
# -------------------------------------------------------------------------
annulus_outer = Wedge((0, 0), 3, 0, 360, 
                      color='blue', 
                      alpha=0.35, 
                      label='Annulus: $2 \leq |z| < 3$')
annulus_inner = Wedge((0, 0), 2, 0, 360, color='white', zorder=2)
ax.add_patch(annulus_outer)
ax.add_patch(annulus_inner)

# Full ring boundary circles
circle_outer = plt.Circle((0, 0), 3, color='dodgerblue', 
                          fill=False, linestyle='--', linewidth=1, zorder=3)
circle_inner = plt.Circle((0, 0), 2, color='dodgerblue', 
                          fill=False, linestyle='-', linewidth=1, zorder=3)
ax.add_patch(circle_outer)
ax.add_patch(circle_inner)

# -------------------------------------------------------------------------
# 2. Whole Sector: -pi/3 <= arg(z) <= pi/2 (Light Red / Pink)
# -------------------------------------------------------------------------
# Plotted out to radius 4 to show it extending beyond the annulus
sector = Wedge((0, 0), 4, -60, 90, 
               color='coral',
               alpha=0.35, zorder=4,
               label=r'Sector: $-\frac{\pi}{3} \leq \arg(z) \leq \frac{\pi}{2}$')
ax.add_patch(sector)

# Sector boundary lines extending out
ax.plot([0, 4 * np.cos(np.radians(-60))], [0, 4 * np.sin(np.radians(-60))], 
        color='firebrick', linestyle=':', linewidth=1.2, zorder=5)
ax.plot([0, 4 * np.cos(np.radians(90))], [0, 4 * np.sin(np.radians(90))], 
        color='firebrick', linestyle=':', linewidth=1.2, zorder=5)

# -------------------------------------------------------------------------
# 3. Intersection Region (Highlighted with bold borders & rich color)
# -------------------------------------------------------------------------
# Solid boundary for inner arc (r=2)
theta_grid = np.linspace(np.radians(-60), np.radians(90), 200)
ax.plot(2 * np.cos(theta_grid), 2 * np.sin(theta_grid), color='darkred', lw=2.5, zorder=6)

# Dashed boundary for outer arc (r=3, strict inequality)
ax.plot(3 * np.cos(theta_grid), 3 * np.sin(theta_grid), color='darkred', lw=2.5, linestyle='--', zorder=6)

# Radial segment boundaries for the intersection region
ax.plot([2 * np.cos(np.radians(-60)), 3 * np.cos(np.radians(-60))],
        [2 * np.sin(np.radians(-60)), 3 * np.sin(np.radians(-60))],
        color='darkred', lw=2.5, zorder=6)
ax.plot([2 * np.cos(np.radians(90)), 3 * np.cos(np.radians(90))],
        [2 * np.sin(np.radians(90)), 3 * np.sin(np.radians(90))], 
        color='darkred', lw=2.5, zorder=6)

# -------------------------------------------------------------------------
# Formatting
# -------------------------------------------------------------------------
ax.set_xlim([-4.2, 4.2])
ax.set_ylim([-4.2, 4.2])
ax.set_xlabel('x')
ax.set_ylabel('y')

# Centered Spines
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

plt.grid(True, linestyle=':', alpha=0.5)
plt.title(r'Intersection of Annulus $2 \leq |z| < 3$ and Sector $-\frac{\pi}{3} \leq \arg(z) \leq \frac{\pi}{2}$', pad=20)

# Custom legend to clarify region overlaps
ax.legend(loc='upper left', frameon=True)

ax.set_aspect('equal', adjustable='box')
plt.tight_layout()

# Save and show
plt.savefig('1_13f_intersection.png')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle

# Create a figure with 3 subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Plot 1: First rectangle (blue)
hr = Rectangle(xy=(-4, -4), 
               width=6, height=8, 
               color='b', alpha=0.5)
vr = Rectangle(xy=(-4, -4), 
               width=8, height=5, 
               color='g', alpha=0.5)

axes[0].plt(hr)
axes[0].add_artist(vr)
axes[0].set_xlim([-5, 5])
axes[0].set_ylim([-5, 5])
axes[0].set_title('Horizontal Rectangle')
axes[0].set_aspect('equal', adjustable='box')
axes[0].grid(True)

# Plot 2: Second rectangle (green)
axes[1].set_xlim([-5, 5])
axes[1].set_ylim([-5, 5])
axes[1].set_title('Vertical Rectangle')
axes[1].set_aspect('equal', adjustable='box')
axes[1].grid(True)

# Plot 3: Circle and Intersection of all shapes
circle = Circle((0, 0), 3, color='r', alpha=0.5, linestyle='-')
axes[2].add_artist(hr)
axes[2].add_artist(vr)
axes[2].add_artist(circle)
axes[2].set_xlim([-5, 5])
axes[2].set_ylim([-5, 5])
axes[2].set_title('Circle & Intersection')
axes[2].set_aspect('equal', adjustable='box')
axes[2].grid(True)

# Set labels for all subplots
for ax in axes:
    ax.set_xlabel('x')
    ax.set_ylabel('y')

# Add overall title and save the figure
plt.suptitle('Two Rectangles, Circle, and Intersection')
plt.tight_layout()
plt.savefig('3_plots_intersection.png')
plt.show()
```
