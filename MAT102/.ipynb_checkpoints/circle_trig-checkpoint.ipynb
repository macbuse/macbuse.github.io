```python
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(111)
ax.set_aspect(1) 

# Move the left and bottom spines to intersect at origin
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')

# Hide redundant borders
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')


ax.text(1, 1, 
        r'$y=x$', 
        fontsize=12, 
        ha='center', 
        va='center'
    )

ax.text(-1, 1, 
        r'$y=-x$', 
        fontsize=12, 
        ha='center', 
        va='center'
    )

# Smooth unit circle
T_circle = np.linspace(0, 2*np.pi, 200)
ax.plot(np.cos(T_circle), np.sin(T_circle),
        color='b', 
        linestyle='--', linewidth=1)

# Defined angles and LaTeX labels from Part 2[cite: 2]
angles_data = [
    (-np.pi/5, r'$-\frac{\pi}{5} = \frac{9\pi}{5} - 2\pi$'),
    (6*np.pi/5, r'$\frac{6\pi}{5}$'),
    (4*np.pi/5, r'$\frac{4\pi}{5}$'),
    (np.pi/2 - np.pi/5, r'$\frac{\pi}{2} - \frac{\pi}{5} = \frac{3\pi}{10}$'),
    (np.pi/5, r'$\frac{\pi}{5} = \frac{\pi}{2} - \frac{3\pi}{10}$'),
    (3*np.pi/2 + np.pi/5, r'$\frac{3\pi}{2} + \frac{\pi}{5} = \frac{17\pi}{10} $')
]

# Plot points and place annotations
for theta, label in angles_data:
    x = np.cos(theta)
    y = np.sin(theta)
    
    # Plot complex number point on unit circle
    ax.scatter(x, y, color='crimson', zorder=5)
    
    # Radius line to point
    ax.plot([0, x], [0, y], color='crimson', linestyle='-', alpha=0.6)
    
    # Position text slightly outside the unit circle (r = 1.15)
    offset_r = 1.15
    ax.text(
        offset_r * np.cos(theta), 
        offset_r * np.sin(theta), 
        label, 
        fontsize=12, 
        ha='center', 
        va='center'
    )

ax.plot([-1,1],[-1,1],'k', linestyle=':', alpha=0.6)
ax.plot([-1,1],[1,-1],'k', linestyle=':', alpha=0.6)

# Set axis limits with margin for labels
ax.set_xlim(-1.3, 1.3)
ax.set_ylim(-1.3, 1.3)

plt.tight_layout()
plt.savefig('exo_8.2.png')
```
