```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# Define the figure and axis
fig, ax = plt.subplots()

# Plot the disc boundary
circle = plt.Circle((0, 0), 3, 
                    color='r', 
                    fill='r', 
                    linestyle='-')

ax.add_artist(circle)

hr = Rectangle(xy = (-4,-4) , 
               width=6, height=8,
               color='b',
               alpha=.5
              )
ax.add_artist(hr)

vr = Rectangle(xy = (-4,-4) , 
               width=8, height=5,
               color='g',
               alpha=.5
              )

ax.add_artist(vr)

# Set axis limits and labels
ax.set_xlim([-4, 4])
ax.set_ylim([-4, 4])
ax.set_xlabel('x')
ax.set_ylabel('y')

# Add grid and title
plt.grid(True)
plt.title('Intersection of $x \leq 2$, $y \leq 1$, and the disc of radius 3')

# Show the plot
plt.gca().set_aspect('equal', adjustable='box')
plt.show
plt.savefig('1_13e.png')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle

# Create a figure with 3 subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Plot 1: First rectangle (blue)
hr = Rectangle(xy=(-4, -4), 
               width=6, height=8, 
               color='b', alpha=0.5)
vr = Rectangle(xy=(-4, -4), 
               width=8, height=5, 
               color='g', alpha=0.5)

axes[0].plt(hr)
axes[0].add_artist(vr)
axes[0].set_xlim([-5, 5])
axes[0].set_ylim([-5, 5])
axes[0].set_title('Horizontal Rectangle')
axes[0].set_aspect('equal', adjustable='box')
axes[0].grid(True)

# Plot 2: Second rectangle (green)
axes[1].set_xlim([-5, 5])
axes[1].set_ylim([-5, 5])
axes[1].set_title('Vertical Rectangle')
axes[1].set_aspect('equal', adjustable='box')
axes[1].grid(True)

# Plot 3: Circle and Intersection of all shapes
circle = Circle((0, 0), 3, color='r', alpha=0.5, linestyle='-')
axes[2].add_artist(hr)
axes[2].add_artist(vr)
axes[2].add_artist(circle)
axes[2].set_xlim([-5, 5])
axes[2].set_ylim([-5, 5])
axes[2].set_title('Circle & Intersection')
axes[2].set_aspect('equal', adjustable='box')
axes[2].grid(True)

# Set labels for all subplots
for ax in axes:
    ax.set_xlabel('x')
    ax.set_ylabel('y')

# Add overall title and save the figure
plt.suptitle('Two Rectangles, Circle, and Intersection')
plt.tight_layout()
plt.savefig('3_plots_intersection.png')
plt.show()
```
